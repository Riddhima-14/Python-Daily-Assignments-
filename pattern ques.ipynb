```python


n=5
for i in range(1,2*n):                       # i=6  row 6 to row 9
    if(i<=n):                                 # i=1  n=5   i<=n  1<=5
        star=i                                 # 1st row no. of star 1
    else:                                      # Mirror
        star=2*n-i
    space=n-star                              # space= 5-1  space=4                 
    
    for j in range(0,space):                   # 4 time      
        print("  ",end="")
        
    for k in range(0,2*star-1):                # (2*1)-1 =1 >>> 1 time        
        print("* ",end="")
        
    print("\n")                      # print()
```
