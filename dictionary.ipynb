```python
# len() :

d={"name":"riddhima","age":22,"course":"MCA AI"}
print(len(d))
```

```python
# keys() :

d={"name":"riddhima","age":22,"course":"MCA AI"}
d.keys()
print(d)
```

```python
# values() :

d={"name":"riddhima","age":22,"course":"MCA AI"}
d.values()
print(d)
```

```python
# items() :

d={"name":"riddhima","age":22,"course":"MCA AI"}
d.items()
print(d)
```

```python
# pop() :

d={"name":"riddhima","age":22,"course":"MCA AI"}
print(d.pop("age"))
```

```python
# clear() :

d={"name":"riddhima","age":22,"course":"MCA AI"}
print(d.clear())
```

```python
# copy() :

d={"name":"riddhima","age":22,"course":"MCA AI"}
data1=d.copy()
print(d)
print(data1)
```

```python
# Membership Operator :

d={"name":"riddhima","age":22,"course":"MCA AI"}
print("age" in d)
print("course"in d)
print("age" not in d)
print("course" not in d)
```
