```python
# Union of two sets

a={1,2,3}
b={3,4,5}
print(a.union(b))
print(a|b)
```

```python
# Intersection of two sets

a={1,2,3,4}
b={3,4,5,6,7}
print(a.intersection(b))
print(a & b)
```

```python
# Difference between two sets

a={1,2,3,4}
b={3,4,5,6}
print(a.difference(b))
print(a-b)
```

```python
# Symmetric difference between two sets

a={1,2,3,4}
b={3,4,5,6,7}
print(a.symmetric_difference(b))
print(a^b)
```

```python
# check subset and superset

a={1,2}
b={1,2,3,4}
print(a.issubset(b))
print(a.issuperset(b))
print(b.issubset(a))
print(b.issuperset(a))
```

```python
# check if two sets are disjoint

a={1,2,3}
b={4,5,6}
print(a.isdisjoint(b))
```

```python
# Set comprehension - squares of numbers
x=1
s={x*x for x in range(1,6)}
print(s)
    
```

```python
# Set comprehension - only even numbers

n=[1,2,3,4,5,6]
even={x for x in n if x%2==0}
print(even)
```

```python

# Count Unique words:

text="my name is Riddhima"

words=text.lower().split()
unique_words= set(words)

print("unique word count: ",len(unique_words))
```

```python
text="my name is Riddhima my name"      # duplicate value is not counted

words=text.lower().split()           #changed in lower case so riddhima and Riddhima both are same....and then split the string by spaceses
unique_words= set(words)             # change the split list into set data type 

print("unique word count: ",len(unique_words))
```
