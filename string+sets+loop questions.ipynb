```python
# count the unique characters in a string:

t="programming"
s=set(t)
c=0
for i in s:
    c+=1
print(c)
```

```python
# print vowels from the string using set:

t="programming"
s=set("aeiou")
for i in t:
    if i in s:
        print(i)
```

```python
# count the vowels :

t="programming"
s=set("aeiou")
c=0
for i in t:
    if i in s:
        c+=1
print(c)
```

```python
# find the duplicate character :

t="programming"
s=set()
d=set()

for i in t:
    if i in s:
        d.add(i)
    else:
        s.add(i)
print(d)
```

```python
# remove the space from the string :

t="programming"
s=set()
for i in t:
    if(i!=" "):
        s.add(i)
print(s)
```
