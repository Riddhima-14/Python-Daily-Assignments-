```python
# len():

a={1,2,3,4,5,6}
print(len(a))
```

```python
# append():

a={1,2,3,4,5,6}
print(a.append(10))
```

```python
# insert():

a={1,2,3,4,5,6}
print(a.insert(10,12))
```

```python
# extend():

a={1,2,3,4,5,6}
print(a.extend(10,12))
```

```python
#remove():

a={1,2,3,4,5,6}
a.remove(1)
print(a)
```

```python
# pop():

a={1,2,3,4,5,6}
x=a.pop()
print(x)

a1={2,3,4,5,6}
print(a1.pop())

a2={4,5,6}
print(a2.pop())
```

```python
# clear():

a={2,3,4,5,6}
a.clear()
print(a)
```

```python
# index():

a={2,3,4,5,6}
a.index(3)
print(a)
```

```python

# count():

a={2,3,4,5,6}
a.count(3)
print(a)
```

```python
# sort():

a={2,3,4,5,6}
a.sort()
print(a)
```

```python

# reverse():

a={2,3,4,5,6}
a.reverse()
print(a)
```

```python

# add() :

a={2,3,4,5,6}
a.add(9)
print(a)
```

```python
# update() :

a={2,3,4,5,6}
a.update([6,7,9])
print(a)
a.update(10)
print(a)
```

```python

# discard() :

a={2,3,4,5,6}
a.discard(4)
print(a)
```

```python

# sum() :

a={2,3,4,5,6}
print(sum(a))
```

```python

# sorted() :

a={3,56,4,46,35,90,56,99}
print(sorted(a))
```

```python
# we can store a tuple in a set: 

a={(1,2,3),(3,6,8),(9,5)}
print(type(a))
print(a)
```
