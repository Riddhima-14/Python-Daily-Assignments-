```python
# add():

d={1,2,3,4,10,5,3,4,7}
d.add(20)
print(d)
```

```python
# update()

d={1,2,3,4,10,5,3,4,7}
d.update([25,40,60])
print(d)
```

```python
# remove()

d={1,2,3,4,10,5,3,4,7}
print(d)
d.remove(10)
#d.remove(55)   ---->>> it will raise an key error because 55 is not present in a set
print(d)
```

```python
# discard()

d={1,2,3,4,10,5,3,4,7}
d1={1,2,3,4,10,5,3,4,7}
print(d)
d.discard(10)
d1.discard(55)   #---->>> it will not raise an any error 
print(d)
print(d1)
```

```python
# pop()

d={40,1,2,3,4,10,5,3,4,7}
print(d)
d.pop()
print(d)
d
```

```python
#clear()

d={40,1,2,3,4,10,5,3,4,7}
print(d)
d.clear()
print(d)
```

```python
# copy()

d={40,1,2,3,4,10,5,3,4,7}
t=d                                # another method to copy a elements
print(t)
d.copy()
print(d)
```

```python
# union()

a={10,20,30}
b={30,40,50}
print(a.union(b))
print(a|b)
```

```python
# intersection()

a={10,20,30}
b={30,40,50}
print(a.intersection(b))
print(a&b)
```

```python
# difference()

a={10,20,30}
b={30,40,50}
print(a.difference(b))
print(a-b)
```
